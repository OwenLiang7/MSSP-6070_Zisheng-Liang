# Step 1a–2e: Upload FAANF stock CSVs, add ticker column, combine all into one DataFrame, and save as 'faanf.csv'.


```python
import pandas as pd
from pathlib import Path

files = ["aapl.csv", "amzn.csv", "fb.csv", "goog.csv", "nflx.csv"]
dfs = []
for fname in files:
    df = pd.read_csv(Path("/content") / fname)
    df["ticker"] = Path(fname).stem.upper()
    dfs.append(df)

combined = pd.concat(dfs, ignore_index=True)
combined.to_csv("/content/faanf.csv", index=False)
print("Saved to /content/faanf.csv; shape:", combined.shape)
combined.head(10)
```

# Step 2f: Data type conversion and sorting

```python
import pandas as pd

df = pd.read_csv("/content/faanf.csv")

# 类型转换
df["date"] = pd.to_datetime(df["date"])
df["volume"] = df["volume"].astype(int)

# 排序
df = df.sort_values(by=["date", "ticker"])

print(df.dtypes)  # 检查类型
df.head()
```

# Step 2h: Use melt() to organize the data

```python
melted = pd.melt(
    df,
    id_vars=["date", "ticker"],
    var_name="variable",
    value_name="value"
)
melted.head(10)
```
